## 1. Python in the age of AI

AI can generate Python code, scripts, and even complete applications in seconds.

So why should we still learn Python?

### AI can write code — but we still need to understand the problem

To use AI effectively, we should be able to:

- define **what we want to achieve**
- design a **workflow** which makes sense
- understand what the individual steps are doing
- read and modify generated code
- recognise when something has gone wrong
- evaluate whether the result actually makes sense

> **AI can write the code, but we still need to know what code we want it to write — and whether the result is doing what we intended.**

### What is the goal of this session?

The goal is **not** to turn you into Python programmers in 2.5 hours.

Instead, we will learn enough Python to:

1. **read** basic Python code
2. **understand** its logic
3. **modify** simple code
4. **recognise and investigate errors**
5. **communicate more effectively with AI coding tools**

For serious work, however, spending additional time learning Python without relying entirely on AI is strongly recommended.

### When something does not work...

A useful workflow is:

**Read the error → identify where the problem occurs → investigate → ask for help if necessary → verify the solution**

AI can be extremely useful for:

- explaining unfamiliar code
- explaining an error message
- suggesting possible solutions
- modifying existing code

But always try to understand **why** the proposed solution works.

## 2. Programming, algorithms and workflows

Before we start writing Python code, we need to know **what we want the computer to do**.

An **algorithm** is a logical sequence of steps designed to perform a task or solve a problem.

A **program** is the implementation of one or more algorithms in a programming language.

### Think of an algorithm as a recipe

A recipe tells us:

1. what inputs we need
2. what steps we should perform
3. in what order
4. what result we expect

Programming works in a similar way.

The computer follows the instructions we provide — it does not understand the archaeological problem by itself.

### Example: calculating artefact density

**Problem:**  
We surveyed 1,200 m² and recorded 356 artefacts.

We want to calculate the number of artefacts per square metre.

**Algorithm:**

1. Define the number of artefacts
2. Define the surveyed area
3. Divide the number of artefacts by the area
4. Report the result

```python
finds = 356
area = 1200

density = finds / area

print(density)
```

The code above is simply an implementation of our algorithm.

At this point, do not worry about the Python syntax.

We will learn what `finds`, `area`, `/`, `=` and `print()` mean in the next sections.

### Computer Vision works in the same way

Suppose we want to detect objects in an archaeological image.

A simplified workflow could be:

1. Load the image
2. Inspect the image
3. Pre-process the image
4. Separate potential objects from the background
5. Detect objects
6. Measure or classify them
7. Inspect and validate the result

**Problem → Algorithm / Workflow → Code → Result → Validation**

This logic will guide the rest of the workshop.

### Where does AI fit in?

AI can help us translate a workflow into code.

But a useful prompt requires us to describe the workflow clearly.

Instead of asking:

> "Detect archaeology in this image."

we should be able to describe:

- what our input data are
- what features we are interested in
- what processing steps may be appropriate
- what output we expect
- how we will evaluate whether the result is correct
