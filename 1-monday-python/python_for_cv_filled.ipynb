### Functions

A function groups reusable instructions into a named block of code.

### Formatting output with f-strings

Python allows us to insert variables directly into text using an **f-string**.

An f-string starts with `f` before the quotation marks, and variables are inserted inside `{}`.

````python
result = calculate_density(356, 1200)

print(f"Artefact density: {result}")
````

Inside an f-string:

- `{result}` inserts the value of the variable
- `{result:.2f}` displays the value with **2 decimal places**

## 4. NumPy

**NumPy** is one of the fundamental libraries for numerical computing in Python.

Its core object is the **array** (`ndarray`).

Arrays allow us to store and process many numerical values efficiently.

This is especially important for Computer Vision because:

> **An image is essentially an array of numbers.**

### Creating a NumPy array

A NumPy array is similar to a Python list, but it is designed for numerical computation.

We create an array using `np.array()`.

````python
import numpy as np

find_counts = np.array([356, 221, 498, 173])

print(find_counts)
print(type(find_counts))
````

Python list        → [356, 221, 498, 173]

NumPy array        → array([356, 221, 498, 173])

### Shape and dimensions

NumPy arrays can have one or more dimensions.

Useful properties include:

- `.shape` — dimensions of the array
- `.size` — total number of elements
- `.ndim` — number of dimensions

````python
# 1D

find_counts = np.array([356, 221, 498, 173])

print("Shape:", find_counts.shape)
print("Size:", find_counts.size)
print("Dimensions:", find_counts.ndim)
````

````python
# 2D

finds_grid = np.array([
    [12, 18, 25],
    [7, 14, 21],
    [3, 9, 16]
])

print(finds_grid)
print("Shape:", finds_grid.shape)
print("Dimensions:", finds_grid.ndim)
````

A 2D array can be interpreted as a grid of values.

For example:

```text
rows × columns
3 × 3

### Indexing NumPy arrays

Just like Python lists, NumPy arrays use indexing starting from `0`.

For a 2D array we specify:

```python
array[row, column]

````python
print(finds_grid)
````

````python
print("Top-left value:", finds_grid[0, 0])
print("Centre value:", finds_grid[1, 1])
print("Bottom-right value:", finds_grid[2, 2])
````

### Slicing

Slicing selects part of an array.

For a 2D array:

```python
array[row_start:row_stop, column_start:column_stop]

````python
# This will later allow us to select or crop parts of an image.

print(finds_grid)

subset = finds_grid[0:2, 0:2]

print(subset)
````

````python
second_column = finds_grid[:, 1]

print(second_column)

````

### Vectorised operations

A major advantage of NumPy is that we can apply an operation to an entire array at once.

We do not need to process each value separately with a loop.

````python
find_counts = np.array([356, 221, 498, 173])

print(find_counts + 100)
````

````python
print(find_counts * 2)
````

````python
print(find_counts)
````

````python
areas = np.array([1200, 900, 1500, 800])

densities = find_counts / areas

print(densities)
````

### Basic statistics

NumPy provides many useful functions for summarising numerical data.

Common examples:

- `np.mean()`
- `np.min()`
- `np.max()`
- `np.sum()`
- `np.std()`

````python
print("Mean:", np.mean(find_counts))
print("Minimum:", np.min(find_counts))
print("Maximum:", np.max(find_counts))
print("Total:", np.sum(find_counts))
````

### Boolean masks

A very important NumPy concept is the **boolean mask**.

When we compare an array with a value, NumPy returns an array of `True` and `False`.

````python
find_counts = np.array([356, 221, 498, 173])

mask = find_counts > 300

print(mask)
````

````python
selected = find_counts[mask]

print(selected)
````

Boolean masks allow us to select values based on a condition.

The same principle is extremely important in image processing.

For example, we can later ask:

> Which pixels are brighter than a selected threshold?

### Data type of an array

NumPy arrays also have a `dtype`, which tells us what type of numerical values they contain.

This becomes very important when working with images.

### From arrays to images

A grayscale image can be represented as a **2D array**:

```text
height × width

A colour image can be represented as a **3D array**:

```text
height × width × channels

For an RGB image, the three channels correspond to:

- Red
- Green
- Blue

````python
image = np.array([
    [0,   50, 100],
    [150, 200, 250],
    [80,  120, 180]
])

print(image)
print("Shape:", image.shape)
````

In this small example, each number can be interpreted as the brightness of one pixel.

For typical 8-bit grayscale images:

- `0` = black
- `255` = white
- values in between = shades of grey

````python
bright_pixels = image > 150

print(bright_pixels)
````

````python
print(image[bright_pixels])
````

We have just performed a very simple form of image segmentation:

> select pixels based on their numerical value.

Later, OpenCV will provide more advanced tools for similar operations.

### Inspect before you trust

Useful questions include:

- What type of object is this?
- What is its shape?
- What is its data type?
- What range of values does it contain?

````python
print("Type:", type(image))
print("Shape:", image.shape)
print("dtype:", image.dtype)
print("Minimum:", image.min())
print("Maximum:", image.max())
````

### Histogram

A histogram shows how frequently different values occur.

For image data, histograms are useful for inspecting the distribution of pixel intensities.

A darker image will usually contain more low pixel values, while a brighter image will contain more high pixel values.

````python
plt.hist(image.flatten(), bins=10)

plt.title("Pixel value distribution")
plt.xlabel("Pixel value")
plt.ylabel("Frequency")

plt.show()
````

Our image is a 2D array, but a histogram expects a single sequence of values.

The `.flatten()` method converts the 2D array into a 1D array.

````python
print("Original shape:", image.shape)
print("Flattened shape:", image.flatten().shape)
````

### Matplotlib can do much more

Matplotlib can also create:

- line plots

- pie charts
- many other scientific visualisations

For this workshop, however, we will mainly use:

- `imshow()` for displaying images

### Visualisation is not the data

Changing the colour map, figure size, title or axes only changes how the data are displayed.

It does **not** change the underlying NumPy array.

This is an important distinction:

> **NumPy stores the data. Matplotlib visualises the data.**

## Displaying an OpenCV image

We can use Matplotlib to display the image.

However, OpenCV and Matplotlib use a different colour-channel order:

- OpenCV: **BGR**
- Matplotlib: **RGB**

If we display the OpenCV image directly, the colours may therefore look incorrect.

````python
plt.imshow(image)
plt.title("OpenCV image displayed directly")
plt.axis("off")
plt.show()
````

We can convert the image from BGR to RGB using:

```python
cv2.cvtColor()
```

````python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Venus of Dolní Věstonice")
plt.axis("off")
plt.show()
````

## Colour channels

A colour image contains three numerical layers.

For an RGB image:

```python
image[:, :, 0]  # Red
image[:, :, 1]  # Green
image[:, :, 2]  # Blue
```

The `:` means: select all rows or columns.

````python
red = image_rgb[:, :, 0]
green = image_rgb[:, :, 1]
blue = image_rgb[:, :, 2]
````

## Converting the image to grayscale

Many Computer Vision operations do not require colour information.

A grayscale image contains only one intensity value per pixel instead of three colour values.

````python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Colour image shape:", image.shape)
print("Grayscale image shape:", gray.shape)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Colour image")
axes[0].axis("off")

axes[1].imshow(gray, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Grayscale image")
axes[1].axis("off")

plt.show()
````

## Pixel values

An image is still just a NumPy array.

We can inspect individual pixel values using indexing.

````python
print("Grayscale pixel:", gray[100, 100])
print("RGB pixel:", image_rgb[100, 100])
````

For a grayscale image, one pixel contains a single intensity value.

For an RGB image, one pixel contains three values:

```text
[R, G, B]
```

For typical 8-bit images:

- `0` = minimum intensity
- `255` = maximum intensity

````python
y = 500
x = 500

plt.imshow(image_rgb)
plt.scatter(x, y)
plt.axis("off")
plt.show()

print("RGB value:", image_rgb[y, x])
````

## Cropping an image

Because an image is a NumPy array, cropping does not require a special OpenCV function.

We simply use NumPy slicing:

```python
image[y_start:y_stop, x_start:x_stop]
```

Notice that the vertical coordinate (`y`) comes first.

````python
crop = image_rgb[150:700, 300:700]

plt.imshow(crop)
plt.title("Cropped region")
plt.axis("off")
plt.show()
````

## Resizing an image

OpenCV provides `cv2.resize()` for changing image dimensions.

The target size is specified as:

```text
(width, height)
```

This is different from NumPy `.shape`, which reports:

```text
(height, width)
```

````python
resized = cv2.resize(image_rgb, (500, 500))

print("Original shape:", image_rgb.shape)
print("Resized shape:", resized.shape)

plt.imshow(resized)
plt.title("Resized image")
plt.axis("off")
plt.show()
````

## Image preprocessing: smoothing

Real images contain texture, small variations and noise.

A common preprocessing step is **smoothing**, which reduces local variations between neighbouring pixels.

One commonly used method is the Gaussian blur.

````python
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray, cmap="gray")
axes[0].set_title("Original grayscale")
axes[0].axis("off")

axes[1].imshow(blurred, cmap="gray")
axes[1].set_title("Gaussian blur")
axes[1].axis("off")

plt.show()
````

The `(5, 5)` parameter defines the size of the neighbourhood used for smoothing.

Larger kernels generally produce stronger smoothing.

The appropriate preprocessing always depends on the image and the analytical goal.

## Thresholding

Our archaeological object is dark, while the background is very bright.

We can therefore try to separate the figurine from the background using a threshold.

The basic idea is:

```text
dark pixels   → object
bright pixels → background
```

This converts the grayscale image into a **binary image**.

````python
threshold_value = 200

_, binary = cv2.threshold(
    blurred,
    threshold_value,
    255,
    cv2.THRESH_BINARY_INV
)
````

We use `THRESH_BINARY_INV` because the figurine is darker than the background.

After thresholding:

- detected object pixels become white
- background pixels become black

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Original")
axes[0].axis("off")

axes[1].imshow(gray, cmap="gray")
axes[1].set_title("Grayscale")
axes[1].axis("off")

axes[2].imshow(binary, cmap="gray")
axes[2].set_title("Binary mask")
axes[2].axis("off")

plt.show()
````

## Connection to NumPy Boolean masks

This is closely related to the Boolean masks we created earlier with NumPy.

For example:

```python
mask = gray < 200
```

and OpenCV thresholding use the same general idea:

> **classify pixels according to their numerical values**

````python
numpy_mask = gray < 200

plt.imshow(numpy_mask, cmap="gray")
plt.title("NumPy Boolean mask")
plt.axis("off")
plt.show()
````

## Detecting the object

Once we have created a binary mask, we can search for connected object boundaries.

OpenCV calls these boundaries **contours**.

````python
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Number of detected contours:", len(contours))
````

The image may contain several small detected regions.

Because the Venus is the dominant object in the image, we can select the **largest contour**.

````python
largest_contour = max(contours, key=cv2.contourArea)

print("Largest contour area:", cv2.contourArea(largest_contour))
````

We can draw the detected contour on a copy of the original image.

````python
detected = image_rgb.copy()

cv2.drawContours(
    detected,
    [largest_contour],
    -1,
    (255, 0, 0),
    3
)

plt.figure(figsize=(7, 7))
plt.imshow(detected)
plt.title("Detected archaeological object")
plt.axis("off")
plt.show()
````

## Bounding box

A detected contour can also be represented by a rectangular bounding box.

````python
x, y, w, h = cv2.boundingRect(largest_contour)

detected_box = image_rgb.copy()

cv2.rectangle(
    detected_box,
    (x, y),
    (x + w, y + h),
    (255, 0, 0),
    3
)

plt.figure(figsize=(7, 7))
plt.imshow(detected_box)
plt.title("Bounding box")
plt.axis("off")
plt.show()
````

The bounding box gives us:

- `x`, `y` — position of the detected object
- `w` — width
- `h` — height

This is already a simple example of turning an image into measurable information.

## From image to measurement

Once an object is detected, we can calculate simple properties such as its area or dimensions.

For example:

````python
object_area = cv2.contourArea(largest_contour)

print(f"Detected object area: {object_area:.0f} pixels²")
print(f"Bounding box width: {w} pixels")
print(f"Bounding box height: {h} pixels")
````

These measurements are expressed in **pixels**.

To convert them into real-world units such as millimetres or centimetres, the image would need to contain spatial calibration information or a scale reference.

## A simple Computer Vision workflow

We have now created a complete basic workflow:

1. load the image
2. inspect the data
3. convert BGR to RGB
4. convert the image to grayscale
5. preprocess the image
6. create a binary mask
7. detect object contours
8. select the archaeological object
9. measure the result
10. visually validate the detection

This is a simple **Computer Vision algorithm**.

````python
# 1. Load image
image = cv2.imread(str(image_path))

# 2. Convert to RGB for visualisation
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 3. Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 4. Smooth the image
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# 5. Create binary mask
_, binary = cv2.threshold(
    blurred,
    200,
    255,
    cv2.THRESH_BINARY_INV
)

# 6. Detect contours
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# 7. Select largest contour
largest_contour = max(contours, key=cv2.contourArea)

# 8. Draw result
result = image_rgb.copy()
cv2.drawContours(result, [largest_contour], -1, (255, 0, 0), 3)
````

````python
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Original")
axes[0].axis("off")

axes[1].imshow(gray, cmap="gray")
axes[1].set_title("Grayscale")
axes[1].axis("off")

axes[2].imshow(binary, cmap="gray")
axes[2].set_title("Binary mask")
axes[2].axis("off")

axes[3].imshow(result)
axes[3].set_title("Detected object")
axes[3].axis("off")

plt.show()
````
